```python
from data_processor import get_city_data, merge_dataframes, ensure_datetime, add_month_column, apply_unit_conversion, aggregate_monthly
from data_visualizer import plot_monthly_sunshine
from config import METRICS

import pandas  as pd
```

```python
df_data_brno = get_city_data("Brno")
df_data_plzen = get_city_data("Plzeň")

df_data_brno.info()
df_data_brno
```

```python
df_data = merge_dataframes([df_data_brno, df_data_plzen])
df_data = ensure_datetime(df_data)
df_data = add_month_column(df_data)

df_data
```

```python
metric_results = []
for metric in METRICS:
    df_metric = apply_unit_conversion(df_data, metric)
    df_metric = aggregate_monthly(df_data, metric)
    metric_results.append(df_metric)

df_data = pd.concat(metric_results, ignore_index=True)
df_data
```

```python
df = df_data.groupby(['city', 'month'], as_index=False).first()
df
```

```python
fig = plot_monthly_sunshine(df)
fig.show()
```
